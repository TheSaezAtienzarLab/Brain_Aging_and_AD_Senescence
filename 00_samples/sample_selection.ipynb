```r
library(tidyverse)
```

```r
meta <- read.csv('NPS-AD_individual_metadata.csv')
meta2 <- read.csv('NPS-AD_neuropathology_and_diagnosis_metadata.csv')
meta3 <- read.csv('NPS-AD_nucleus_hashing_metadata.csv')
```

```r
head(meta)
head(meta2)
head(meta3)
```

```r
# Step 1: Merge
merged_meta <- merge(meta, meta2, by.x = "individualID", by.y = "IndividualID")
```

```r
head(merged_meta)
```

```r
# Check column names
names(merged_meta)
```

```r
# Remove all columns ending in .x (keep .y version from the second dataset)
merged_meta_clean <- merged_meta[, !grepl("\\.x$", names(merged_meta))]

# Optionally, rename .y columns back to original names for clarity
names(merged_meta_clean) <- sub("\\.y$", "", names(merged_meta_clean))

# View cleaned column names
names(merged_meta_clean)
```

```r
diagnosis_cols <- c("CERAD", "CDR", "BRAAK_AD", "PD", "FTD", 
                    "MS", "SCZ", "BD", "PTSD", "ADHD", "OCD")
```

```r
healthy_controls <- merged_meta_clean[
  apply(merged_meta_clean[, diagnosis_cols, drop = FALSE], 1, function(row) all(is.na(row) | row == 0)),
]
```

```r
healthy_controls <- healthy_controls[!is.na(healthy_controls$ageDeath), ]

dim(healthy_controls)
head(healthy_controls)
```

```r
names(healthy_controls)
str(healthy_controls)
```

```r
# Step 1: Clean column names (optional standardization)
names(healthy_controls) <- gsub(" ", "_", names(healthy_controls))  # Replace spaces with underscores
names(healthy_controls) <- gsub("[^[:alnum:]_]", "", names(healthy_controls))  # Remove non-alphanumerics

# Step 2: Convert ageDeath to numeric (it's currently a character)
healthy_controls$ageDeath <- as.numeric(healthy_controls$ageDeath)

# Step 3: Convert yearsEducation from logical to integer (if needed)
if (is.logical(healthy_controls$yearsEducation)) {
  healthy_controls$yearsEducation <- as.integer(NA)
}

summary(healthy_controls$ageDeath)
hist(healthy_controls$ageDeath, breaks = 30, main = "Histogram of Age at Death", xlab = "Age at Death")
```

```r
# Step 1: Remove individuals with ageDeath <= 0 or NA
healthy_controls <- healthy_controls[!is.na(healthy_controls$ageDeath) & healthy_controls$ageDeath > 0, ]

# Step 2: Convert ageDeath to numeric (if not already)
healthy_controls$ageDeath <- as.numeric(healthy_controls$ageDeath)

# Step 3: Define breaks for decade-based age groups
max_age <- ceiling(max(healthy_controls$ageDeath, na.rm = TRUE))
age_breaks <- seq(0, max_age + 10, by = 10)  # e.g., 0, 10, 20, ..., up to max_age

# Step 4: Create labels like "0-9 y/o", "10-19 y/o", etc.
age_labels <- paste0(age_breaks[-length(age_breaks)], "-", age_breaks[-1] - 1, " y/o")

# Step 5: Add age_group column
healthy_controls$age_group <- cut(
  healthy_controls$ageDeath,
  breaks = age_breaks,
  labels = age_labels,
  right = FALSE,
  include.lowest = TRUE
)

# Step 6: Check result
table(healthy_controls$age_group)
```

```r
# First, standardize column names to match:
names(meta3)[names(meta3) == "IndividualID"] <- "individualID"

# Merge: inner join on individualID, keep only matches
healthy_controls_merged <- merge(healthy_controls, meta3, by = "individualID")

# Check the merged result
dim(healthy_controls_merged)
head(healthy_controls_merged)
```

```r
write.csv(healthy_controls_merged, "healthy_controls.csv", row.names = FALSE)
```
