# Figure 4A

```python
META_PATH = 'results/fig4A_source_data.csv'

ARM = dict(
    title="Susceptibility (AD arm)",
    xlabel="odds ratio, AD vs Control",
    formula="is_senescent ~ Study_Group + log10_total_counts + Sex + age_dec + (1|Donor)",
    method="cell-level logistic GLMM within cell type, depth adjusted",
    note="Odds of a cell of that type being senescent in AD, relative to "
         "age-matched controls.",
)
CONTRAST  = "AD vs Control"
MAJOR_CTS = ["Microglia", "Astrocyte", "Oligodendrocyte", "OPC",
             "Excitatory", "Inhibitory"]
COHORT_MARK = {"PsychAD": "s", "PsychEncode": "o", "PsychENCODE": "o", "Mathys": "o"}

import ast, numpy as np, pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon
from matplotlib.lines import Line2D

def fmt_b(v):
    return f"{v:.3f}"

def fmt_ci(lo, hi):
    return f"[{lo:.3f}, {hi:.3f}]"

def fmt_p(v):
    return f"{v:.1e}" if v < 0.001 else (f"{v:.3f}" if v < 0.01 else f"{v:.2f}")

meta = pd.read_csv(META_PATH)
as_list = lambda v: v if isinstance(v, (list, tuple, np.ndarray)) else ast.literal_eval(str(v))
for c in ("cohort_names", "cohort_betas", "cohort_ses"):
    meta[c] = meta[c].map(as_list)
if 'pooled_se' not in meta.columns:
    meta['pooled_se'] = (meta.ci_hi - meta.ci_lo) / (2 * 1.96)
if 'OR' not in meta.columns:
    meta['OR']    = np.exp(meta['pooled_beta'])
    meta['OR_lo'] = np.exp(meta['ci_lo'])
    meta['OR_hi'] = np.exp(meta['ci_hi'])
    print("derived OR, OR_lo, OR_hi from the beta columns")
meta = meta[meta.Cell_Type.isin(MAJOR_CTS)]
meta = (meta.assign(_o=meta.Cell_Type.map({c: i for i, c in enumerate(MAJOR_CTS)}))
            .sort_values("_o").drop(columns="_o").reset_index(drop=True))

names = list(dict.fromkeys(n for r in meta.cohort_names for n in r))
n = len(meta)

x = {"ct": 0.005, "fL": 0.15, "fR": 0.42}
for i, nm in enumerate(names): x[nm] = 0.46 + i * 0.15
b0 = 0.46 + len(names) * 0.15
cols = [("pooled", b0, "center"), ("p", b0 + 0.15, "right")]
x.update({k: v for k, v, _ in cols})
heads = {"pooled": "Pooled OR", "p": "p"}

vals = [v for _, r in meta.iterrows() for v in list(r.cohort_betas) + [r.pooled_beta]]
lo, hi = min(vals), max(vals)
pad = max((hi - lo) * 0.5, 1e-4); lo, hi = lo - pad, hi + pad
to_x = lambda v: x["fL"] + (np.clip(v, lo, hi) - lo) / (hi - lo) * (x["fR"] - x["fL"])
off = [-0.15, 0.15] if len(names) == 2 else list(np.linspace(-0.18, 0.18, len(names)))

fig = plt.figure(figsize=(9.8, max(3.4, n * 0.64 + 2.6)))
ax = fig.add_subplot(111)
y_hdr, y_lbl = n - 0.5, n - 0.2

for nm in names:
    ax.text(x[nm], y_lbl, nm, fontsize=7.5, style="italic",
            color=color_for('cohort', [nm])[0], ha="center", va="bottom")
    ax.text(x[nm], y_hdr, "OR (95% CI)", fontsize=9, fontweight="bold",
            ha="center", va="bottom")
ax.text(x["ct"], y_hdr, "Cell Type", fontsize=9, fontweight="bold", ha="left", va="bottom")
for key, xx, align in cols:
    ax.text(xx, y_hdr, heads[key], fontsize=9, fontweight="bold", ha=align, va="bottom")
ax.plot([0, 0.98], [y_hdr - 0.10, y_hdr - 0.10], "k-", lw=1, clip_on=False)

for i, r in meta.iterrows():
    yy = n - i - 1
    if i % 2 == 0:
        ax.axhspan(yy - 0.5, yy + 0.5, xmin=0.004, xmax=0.996, color="#f7f7f7", zorder=0)
    c = color_for('cell_type', [r.Cell_Type])[0]
    sigp = r.p < 0.05
    ax.text(x["ct"], yy, ("* " if sigp else "") + r.Cell_Type,
            fontsize=9, fontweight="bold" if sigp else "normal",
            ha="left", va="center", color=c if sigp else "#333")

    for j, (nm, bb, se) in enumerate(zip(r.cohort_names, r.cohort_betas, r.cohort_ses)):
        yo = yy + off[j]; cc = color_for('cohort', [nm])[0]
        ax.plot([to_x(bb - 1.96 * se), to_x(bb + 1.96 * se)], [yo, yo],
                color=cc, lw=1.4, alpha=0.8, zorder=2)
        ax.plot(to_x(bb), yo, COHORT_MARK.get(nm, "o"), ms=5.5,
                color=cc, mec="black", mew=0.4, zorder=3)
        ax.text(x[nm], yy + 0.13, fmt_b(np.exp(bb)), fontsize=7.8, ha="center", va="bottom")
        ax.text(x[nm], yy - 0.13, fmt_ci(np.exp(bb - 1.96 * se), np.exp(bb + 1.96 * se)),
                fontsize=6.5, ha="center", va="top", color="#666")

    xr = to_x(r.pooled_beta)
    ax.plot([to_x(r.ci_lo), to_x(r.ci_hi)], [yy, yy], color=c, lw=2.2, alpha=0.45, zorder=4)
    ax.add_patch(Polygon([[xr - 0.008, yy], [xr, yy - 0.17], [xr + 0.008, yy],
                          [xr, yy + 0.17]], fc=c, ec="black", lw=0.5, zorder=5, alpha=0.9))

    cell = {"pooled": (fmt_b(r.OR), fmt_ci(r.OR_lo, r.OR_hi)),
            "p":      (fmt_p(r.p), None)}
    for key, xx, align in cols:
        top, bot = cell[key]
        fw = "bold" if sigp else "normal"
        col = c if (key == "p" and sigp) else "#333"
        ax.text(xx, yy + (0.13 if bot else 0), top, fontsize=7.8, ha=align,
                va="bottom" if bot else "center", fontweight=fw, color=col)
        if bot:
            ax.text(xx, yy - 0.13, bot, fontsize=6.5, ha=align, va="top", color="#666")

ya = -0.55
ax.plot([x["fL"], x["fR"]], [ya, ya], "k-", lw=1)
ax.axvline(to_x(0), color="#888", ls="--", lw=0.8, ymin=0.06, ymax=0.92, alpha=0.6)
for t in np.linspace(lo, hi, 6):
    ax.plot([to_x(t)] * 2, [ya, ya - 0.09], "k-", lw=0.8)
    ax.text(to_x(t), ya - 0.15, f"{np.exp(t):.2f}",
            fontsize=7.2, ha="center", va="top")
ax.text((x["fL"] + x["fR"]) / 2, ya - 0.52, ARM["xlabel"],
        fontsize=8.5, fontweight="bold", ha="center")

leg = [Line2D([0], [0], marker=COHORT_MARK.get(nm, "o"), color="w",
              markerfacecolor=color_for('cohort', [nm])[0], markersize=5,
              markeredgecolor="black", markeredgewidth=0.3, label=nm) for nm in names]
leg.append(Line2D([0], [0], marker="D", color="w", markerfacecolor="#666",
                  markersize=6, markeredgecolor="black", label="Pooled (RE)"))
ax.legend(handles=leg, loc="upper center",
          bbox_to_anchor=((x["fL"] + x["fR"]) / 2, ya - 0.72),
          bbox_transform=ax.transData, fontsize=7.5, frameon=False,
          ncol=len(leg), handletextpad=0.35, columnspacing=1.1)

ax.text(0.5, y_lbl + 0.45,
        f"{ARM['title']} meta-analysis — {CONTRAST} ({len(names)} cohorts)",
        fontsize=10, fontweight="bold", ha="center")
ax.text(0.005, ya - 1.12,
        f"* p<0.05   ·   {ARM['note']}",
        fontsize=7.3, ha="left", va="top", color="#666")
ax.text(0.005, ya - 1.42,
        f"Model: {ARM['formula']}   ·   {ARM['method']}   ·   "
        f"DerSimonian-Laird random effects",
        fontsize=7.0, ha="left", va="top", color="#555", style="italic")

ax.set_xlim(-0.01, 1.00); ax.set_ylim(-2.35, y_lbl + 0.75); ax.axis("off")
plt.tight_layout()
plt.show()
```

# Figure 4B

```python
SRC         = 'results/fig4B_source_data.csv'
GROUP_ORDER = ['YHC', 'HC', 'AD']
GROUP_LABEL = {'YHC': 'Young HC', 'HC': 'HC', 'AD': 'AD'}
CELL_ORDER  = ['Excitatory', 'Inhibitory', 'Astrocyte', 'Oligodendrocyte', 'Microglia', 'OPC']
BAR_LABELS  = True

import numpy as np, pandas as pd, matplotlib.pyplot as plt

d = pd.read_csv(SRC)
d.columns = [c.strip() for c in d.columns]
d = d.loc[:, ~d.columns.str.match(r'^Unnamed')]

h = d.pivot(index='Group', columns='Cell_Type', values='Susceptibility_%')
miss = [c for c in CELL_ORDER if c not in h.columns] + [g for g in GROUP_ORDER if g not in h.index]
if miss:
    raise ValueError(f"not in file: {miss}")
h = h.reindex(index=GROUP_ORDER, columns=CELL_ORDER)

ct_hex = color_for('cell_type', CELL_ORDER)
print("Fig4B colors:", dict(zip(CELL_ORDER, ct_hex)))

n_ct = len(CELL_ORDER)
x    = np.arange(len(GROUP_ORDER))
w    = 0.78 / n_ct
ymax = float(np.nanmax(h.values))

fig, ax = plt.subplots(figsize=(4.6, 3.2))
for k, ct in enumerate(CELL_ORDER):
    xx   = x + (k - (n_ct - 1) / 2) * w
    v    = h[ct].values
    bars = ax.bar(xx, v, width=w * 0.9, color=ct_hex[k],
                  edgecolor='white', linewidth=0.3, label=ct, zorder=2)
    if BAR_LABELS:
        for bar, val in zip(bars, v):
            ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + ymax * 0.015,
                    f'{val:.1f}', ha='center', va='bottom', fontsize=4, rotation=90)

ax.set_xticks(x)
ax.set_xticklabels([GROUP_LABEL.get(g, g) for g in GROUP_ORDER], fontsize=8)
ax.set_ylabel('Senescence fraction within cell type (%)', fontsize=8)
ax.set_ylim(0, ymax * 1.22)
ax.tick_params(labelsize=8)
for s in ['top', 'right']:
    ax.spines[s].set_visible(False)
ax.legend(fontsize=6.5, frameon=False, ncol=2, loc='upper left',
          handlelength=0.9, handletextpad=0.4, columnspacing=0.8, labelspacing=0.3)
fig.tight_layout()
plt.show()
```

# Figure 4C

```python
SRC          = 'results/fig4C_source_data.csv'
GROUP_ORDER  = ['YHC', 'HC', 'AD']
FACET_ORDER  = ['Microglia', 'Astrocyte', 'Oligodendrocyte',
                'Excitatory', 'OPC', 'Inhibitory']
NROW, NCOL   = 2, 3
GROUP_MAP    = {}
YMAX         = 30
JITTER_SEED  = 42        # same as Fig 1D
JITTER_W     = 0.06      # same as Fig 1D
DOT_DARKEN   = -0.45     # how much darker the points are than the box

import numpy as np, pandas as pd, matplotlib.pyplot as plt

d = pd.read_csv(SRC)
d.columns = [c.strip() for c in d.columns]
d = d.loc[:, ~d.columns.str.match(r'^Unnamed')]
miss = ([c for c in FACET_ORDER if c not in set(d.Cell_Type)] +
        [g for g in GROUP_ORDER if g not in set(d.Group)])
if miss:
    raise ValueError(f"not in file: {miss}")

lvls  = [GROUP_MAP.get(g, g) for g in GROUP_ORDER]
g_hex = color_for('study_group', lvls)
print("Fig4C colors:", dict(zip(GROUP_ORDER, g_hex)))
rng = np.random.default_rng(JITTER_SEED)

def shade(hexcol, f):
    """f<0 darkens toward black, f>0 lightens toward white."""
    r, g, b = (int(hexcol[i:i + 2], 16) for i in (1, 3, 5))
    t = 0 if f < 0 else 255
    k = abs(f)
    return "#%02x%02x%02x" % tuple(int(round(c + (t - c) * k)) for c in (r, g, b))

dot_hex = [shade(h, DOT_DARKEN) for h in g_hex]

fig, axes = plt.subplots(NROW, NCOL, figsize=(2.3 * NCOL, 2.35 * NROW),
                         sharex=True, sharey=True)
axes = np.atleast_1d(axes).ravel()

for ax, ct in zip(axes, FACET_ORDER):
    sub = d[d.Cell_Type == ct]
    series = [sub.loc[sub.Group == g, 'Pct_senescent'].dropna().values
              for g in GROUP_ORDER]
    bp = ax.boxplot(series, positions=np.arange(len(GROUP_ORDER)), widths=0.6,
                    showfliers=False, patch_artist=True, zorder=2)
    for patch, hx in zip(bp['boxes'], g_hex):
        patch.set(facecolor=hx, alpha=1.0, edgecolor=shade(hx, -0.35),
                  linewidth=0.9, zorder=2)
    for key, hx in (('whiskers', np.repeat(g_hex, 2)), ('caps', np.repeat(g_hex, 2))):
        for art, h in zip(bp[key], hx):
            art.set(color=shade(h, -0.35), linewidth=0.9, zorder=2)
    for art in bp['medians']:
        art.set(color='black', linewidth=1.2, zorder=4)
    for i, (v, dx) in enumerate(zip(series, dot_hex)):
        ax.scatter(rng.normal(i, JITTER_W, len(v)), v, s=9, color=dx,
                   alpha=0.55, edgecolors='white', linewidth=0.2, zorder=3,
                   rasterized=True)
    ax.set_title(ct, fontsize=8, fontweight='bold')
    ax.set_ylim(0, YMAX)
    ax.tick_params(labelsize=7)
    for s in ['top', 'right']:
        ax.spines[s].set_visible(False)

for ax in axes[len(FACET_ORDER):]:
    ax.set_visible(False)
for ax in axes:
    if ax.get_visible():
        ax.set_xticks(np.arange(len(GROUP_ORDER)))
        ax.set_xticklabels(GROUP_ORDER, fontsize=7)
        ax.tick_params(axis='x', labelbottom=True)
for i, ax in enumerate(axes[:len(FACET_ORDER)]):
    if i % NCOL == 0:
        ax.set_ylabel('SnC (%)', fontsize=8)

fig.tight_layout()
plt.show()
```

```python
META_PATH = 'results/fig4D_source_data.csv'

ARM = dict(
    title="Burden (AD arm)",
    xlabel="odds ratio, AD vs Control",
    formula="is_burden ~ Study_Group + ct_prop_z + log10_total_counts + Sex + age_dec + (1|Donor)",
    method="cell-level logistic GLMM, ct_prop + per-cell depth adjusted",
    note="Odds of a cell being a senescent cell of that type, out of all of that "
         "donor's cells, in AD relative to age-matched controls.",
)
CONTRAST  = "AD vs Control"
MAJOR_CTS = ["Oligodendrocyte", "Microglia", "OPC", "Astrocyte",
             "Excitatory", "Inhibitory"]
COHORT_MARK = {"PsychAD": "s", "PsychEncode": "o", "PsychENCODE": "o", "Mathys": "o"}

import ast, numpy as np, pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon
from matplotlib.lines import Line2D

def fmt_b(v):
    return f"{v:.3f}"

def fmt_ci(lo, hi):
    return f"[{lo:.3f}, {hi:.3f}]"

def fmt_p(v):
    return f"{v:.1e}" if v < 0.001 else (f"{v:.3f}" if v < 0.01 else f"{v:.2f}")

meta = pd.read_csv(META_PATH)
as_list = lambda v: v if isinstance(v, (list, tuple, np.ndarray)) else ast.literal_eval(str(v))
for c in ("cohort_names", "cohort_betas", "cohort_ses"):
    meta[c] = meta[c].map(as_list)
if 'pooled_se' not in meta.columns:
    meta['pooled_se'] = (meta.ci_hi - meta.ci_lo) / (2 * 1.96)
if 'OR' not in meta.columns:
    meta['OR']    = np.exp(meta['pooled_beta'])
    meta['OR_lo'] = np.exp(meta['ci_lo'])
    meta['OR_hi'] = np.exp(meta['ci_hi'])
    print("derived OR, OR_lo, OR_hi from the beta columns")
meta = meta[meta.Cell_Type.isin(MAJOR_CTS)]
meta = (meta.assign(_o=meta.Cell_Type.map({c: i for i, c in enumerate(MAJOR_CTS)}))
            .sort_values("_o").drop(columns="_o").reset_index(drop=True))

names = list(dict.fromkeys(n for r in meta.cohort_names for n in r))
n = len(meta)

x = {"ct": 0.005, "fL": 0.15, "fR": 0.42}
for i, nm in enumerate(names): x[nm] = 0.46 + i * 0.15
b0 = 0.46 + len(names) * 0.15
cols = [("pooled", b0, "center"), ("p", b0 + 0.15, "right")]
x.update({k: v for k, v, _ in cols})
heads = {"pooled": "Pooled OR", "p": "p"}

vals = [v for _, r in meta.iterrows() for v in list(r.cohort_betas) + [r.pooled_beta]]
lo, hi = min(vals), max(vals)
pad = max((hi - lo) * 0.5, 1e-4); lo, hi = lo - pad, hi + pad
to_x = lambda v: x["fL"] + (np.clip(v, lo, hi) - lo) / (hi - lo) * (x["fR"] - x["fL"])
off = [-0.15, 0.15] if len(names) == 2 else list(np.linspace(-0.18, 0.18, len(names)))

fig = plt.figure(figsize=(9.8, max(3.4, n * 0.64 + 2.6)))
ax = fig.add_subplot(111)
y_hdr, y_lbl = n - 0.5, n - 0.2

for nm in names:
    ax.text(x[nm], y_lbl, nm, fontsize=7.5, style="italic",
            color=color_for('cohort', [nm])[0], ha="center", va="bottom")
    ax.text(x[nm], y_hdr, "OR (95% CI)", fontsize=9, fontweight="bold",
            ha="center", va="bottom")
ax.text(x["ct"], y_hdr, "Cell Type", fontsize=9, fontweight="bold", ha="left", va="bottom")
for key, xx, align in cols:
    ax.text(xx, y_hdr, heads[key], fontsize=9, fontweight="bold", ha=align, va="bottom")
ax.plot([0, 0.98], [y_hdr - 0.10, y_hdr - 0.10], "k-", lw=1, clip_on=False)

for i, r in meta.iterrows():
    yy = n - i - 1
    if i % 2 == 0:
        ax.axhspan(yy - 0.5, yy + 0.5, xmin=0.004, xmax=0.996, color="#f7f7f7", zorder=0)
    c = color_for('cell_type', [r.Cell_Type])[0]
    sigp = r.p < 0.05
    ax.text(x["ct"], yy, ("* " if sigp else "") + r.Cell_Type,
            fontsize=9, fontweight="bold" if sigp else "normal",
            ha="left", va="center", color=c if sigp else "#333")

    for j, (nm, bb, se) in enumerate(zip(r.cohort_names, r.cohort_betas, r.cohort_ses)):
        yo = yy + off[j]; cc = color_for('cohort', [nm])[0]
        ax.plot([to_x(bb - 1.96 * se), to_x(bb + 1.96 * se)], [yo, yo],
                color=cc, lw=1.4, alpha=0.8, zorder=2)
        ax.plot(to_x(bb), yo, COHORT_MARK.get(nm, "o"), ms=5.5,
                color=cc, mec="black", mew=0.4, zorder=3)
        ax.text(x[nm], yy + 0.13, fmt_b(np.exp(bb)), fontsize=7.8, ha="center", va="bottom")
        ax.text(x[nm], yy - 0.13, fmt_ci(np.exp(bb - 1.96 * se), np.exp(bb + 1.96 * se)),
                fontsize=6.5, ha="center", va="top", color="#666")

    xr = to_x(r.pooled_beta)
    ax.plot([to_x(r.ci_lo), to_x(r.ci_hi)], [yy, yy], color=c, lw=2.2, alpha=0.45, zorder=4)
    ax.add_patch(Polygon([[xr - 0.008, yy], [xr, yy - 0.17], [xr + 0.008, yy],
                          [xr, yy + 0.17]], fc=c, ec="black", lw=0.5, zorder=5, alpha=0.9))

    cell = {"pooled": (fmt_b(r.OR), fmt_ci(r.OR_lo, r.OR_hi)),
            "p":      (fmt_p(r.p), None)}
    for key, xx, align in cols:
        top, bot = cell[key]
        fw = "bold" if sigp else "normal"
        col = c if (key == "p" and sigp) else "#333"
        ax.text(xx, yy + (0.13 if bot else 0), top, fontsize=7.8, ha=align,
                va="bottom" if bot else "center", fontweight=fw, color=col)
        if bot:
            ax.text(xx, yy - 0.13, bot, fontsize=6.5, ha=align, va="top", color="#666")

ya = -0.55
ax.plot([x["fL"], x["fR"]], [ya, ya], "k-", lw=1)
ax.axvline(to_x(0), color="#888", ls="--", lw=0.8, ymin=0.06, ymax=0.92, alpha=0.6)
for t in np.linspace(lo, hi, 6):
    ax.plot([to_x(t)] * 2, [ya, ya - 0.09], "k-", lw=0.8)
    ax.text(to_x(t), ya - 0.15, f"{np.exp(t):.2f}",
            fontsize=7.2, ha="center", va="top")
ax.text((x["fL"] + x["fR"]) / 2, ya - 0.52, ARM["xlabel"],
        fontsize=8.5, fontweight="bold", ha="center")

leg = [Line2D([0], [0], marker=COHORT_MARK.get(nm, "o"), color="w",
              markerfacecolor=color_for('cohort', [nm])[0], markersize=5,
              markeredgecolor="black", markeredgewidth=0.3, label=nm) for nm in names]
leg.append(Line2D([0], [0], marker="D", color="w", markerfacecolor="#666",
                  markersize=6, markeredgecolor="black", label="Pooled (RE)"))
ax.legend(handles=leg, loc="upper center",
          bbox_to_anchor=((x["fL"] + x["fR"]) / 2, ya - 0.72),
          bbox_transform=ax.transData, fontsize=7.5, frameon=False,
          ncol=len(leg), handletextpad=0.35, columnspacing=1.1)

ax.text(0.5, y_lbl + 0.45,
        f"{ARM['title']} meta-analysis — {CONTRAST} ({len(names)} cohorts)",
        fontsize=10, fontweight="bold", ha="center")
ax.text(0.005, ya - 1.12,
        f"* p<0.05   ·   {ARM['note']}",
        fontsize=7.3, ha="left", va="top", color="#666")
ax.text(0.005, ya - 1.42,
        f"Model: {ARM['formula']}   ·   {ARM['method']}   ·   "
        f"DerSimonian-Laird random effects",
        fontsize=7.0, ha="left", va="top", color="#555", style="italic")

ax.set_xlim(-0.01, 1.00); ax.set_ylim(-2.35, y_lbl + 0.75); ax.axis("off")
plt.tight_layout()
plt.show()
```

# Figure 4E

```python
SRC         = 'results/fig4E_source_data.csv'
GROUP_ORDER = ['YHC', 'HC', 'AD']
GROUP_LABEL = {'YHC': 'Young HC', 'HC': 'HC', 'AD': 'AD'}
CELL_ORDER  = ['Excitatory', 'Inhibitory', 'Astrocyte', 'Oligodendrocyte', 'Microglia', 'OPC']
NORMALIZE   = False      # False reproduces page 12. True gives the 0-100 stack
SHOW_TOTALS = True

import numpy as np, pandas as pd, matplotlib.pyplot as plt

d = pd.read_csv(SRC)
d.columns = [c.strip() for c in d.columns]
d = d.loc[:, ~d.columns.str.match(r'^Unnamed')]

VALUE = 'Burden_share_%' if NORMALIZE else 'Burden_%'
if VALUE not in d.columns:
    raise KeyError(f"{VALUE} not in file. columns: {list(d.columns)}")
b = d.pivot(index='Group', columns='Cell_Type', values=VALUE)
miss = [c for c in CELL_ORDER if c not in b.columns] + [g for g in GROUP_ORDER if g not in b.index]
if miss:
    raise ValueError(f"not in file: {miss}")
b = b.reindex(index=GROUP_ORDER, columns=CELL_ORDER)

totals = b.sum(axis=1)
print(f"{VALUE} group totals:", totals.round(3).to_dict())

ct_hex = color_for('cell_type', CELL_ORDER)
print("Fig4E colors:", dict(zip(CELL_ORDER, ct_hex)))

x, bottom = np.arange(len(GROUP_ORDER)), np.zeros(len(GROUP_ORDER))
top = float(totals.max())

fig, ax = plt.subplots(figsize=(3.6, 3.4))
for k, ct in enumerate(CELL_ORDER):
    v = b[ct].values
    ax.bar(x, v, bottom=bottom, width=0.6, color=ct_hex[k],
           edgecolor='white', linewidth=0.5, label=ct, zorder=2)
    bottom = bottom + v

if NORMALIZE:
    ax.set_ylabel('Contribution to senescent pool (%)', fontsize=8.5)
    ax.set_ylim(0, 100); ax.set_yticks([0, 25, 50, 75, 100])
else:
    ax.set_ylabel('Senescence burden\n(% of all cells among that group)', fontsize=8.5)
    ax.set_ylim(0, top * 1.16)
if SHOW_TOTALS:
    for xi, t in zip(x, totals.values):
        ax.text(xi, t + top * 0.02, f'{t:.2f}%', ha='center', va='bottom',
                fontsize=7.5, fontweight='bold', color='#333')

ax.set_xticks(x)
ax.set_xticklabels([GROUP_LABEL.get(g, g) for g in GROUP_ORDER], fontsize=8)
ax.tick_params(labelsize=8)
for s in ['top', 'right']:
    ax.spines[s].set_visible(False)
ax.legend(fontsize=6.5, frameon=False, loc='center left',
          bbox_to_anchor=(1.01, 0.5), handlelength=0.9, handletextpad=0.4)
fig.tight_layout()
plt.show()
```
